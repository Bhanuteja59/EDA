```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# https://www.kaggle.com/datasets/unsdsn/world-happiness?resource=download


data = pd.read_csv("2017.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
for i in data.select_dtypes(np.number).columns:
  sns.histplot(data[i],kde=True)
  plt.show()
```

```python
for i in data.select_dtypes(include=["int","float"]).columns :
  sns.boxplot(data[i])
  plt.show()
```

```python
key_features = ["Happiness.Score", "Economy..GDP.per.Capita.", "Health..Life.Expectancy.", "Freedom"]
```

```python
for i in key_features:
  sns.histplot(data[i])
  plt.show()
```

```python
for i in data[key_features]:
    sns.scatterplot(data=data, x=i, y="Happiness.Score")
    plt.show()
```

```python
corr_features = ["Happiness.Score", "Economy..GDP.per.Capita.", 
                 "Health..Life.Expectancy.", "Trust..Government.Corruption."]
```

```python
sns.heatmap(data[corr_features].corr())
plt.title("Correlation Heatmap: Happiness vs Key Factors", fontsize=14)
plt.show()
```
